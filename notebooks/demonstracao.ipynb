Ferramenta para conversao de arquivos de dados tabulares entre **CSV**, **TXT**, **XLSX**, **Parquet** e **JSON**. Inclui deteccao automatica de delimitador e codificacao (encoding), conversao em lote e validacao de integridade estrutural.

## 1. Instalacao e Preparacao do Ambiente
Execute as celulas abaixo para carregar as dependencias necessarias:

```python
# Se estiver executando no Google Colab pela primeira vez:
# !git clone https://github.com/suelio/conversor_arquivos.git
# %cd conversor_arquivos

import pandas as pd
from conversor import FileConverter, FileReader

print('Ambiente pronto e FileConverter importado com sucesso.')
```

## 2. Inspecao e Leitura Inteligente
O modulo `FileReader` detecta autonomamente a codificacao e o delimitador, mesmo em arquivos com padrao regional (ex: ponto e virgula):

```python
# Lendo arquivo de exemplo e exibindo metadados
df, meta = FileReader.read('data/exemplo_vendas.csv')
print('Metadados identificados:', meta)
display(df.head())
```

## 3. Conversao Individual com Validacao e Metricas
Convertendo de CSV para Parquet (otimizado para processamento em larga escala):

```python
resultado = FileConverter.convert(
    source='data/exemplo_vendas.csv',
    target_format='parquet',
    validate=True
)

for k, v in resultado.items():
    print(f"{k}: {v}")
```

## 4. Conversao Cruzada: Parquet para Excel (.xlsx)

```python
res_excel = FileConverter.convert(
    source='data/exemplo_vendas.parquet',
    target_format='xlsx',
    validate=True
)
print(f"Arquivo gerado: {res_excel['arquivo_destino']} com {res_excel['linhas']} linhas.")
```

## 5. Conversao em Lote (Diretorios Inteiros)
Processa todos os arquivos compativeis de uma pasta de uma so vez, consolidando o resultado em um sumario executivo:

```python
res_lote = FileConverter.convert_batch(
    source_dir='data',
    target_format='json',
    output_dir='data/saida_json'
)

print(f"Total processado: {res_lote['total_arquivos']} arquivos")
print(f"Sucessos: {res_lote['sucessos']} | Falhas: {res_lote['falhas']}")
print(f"Tempo de execucao: {res_lote['tempo_total_segundos']} segundos")
```
